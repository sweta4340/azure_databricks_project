```python
df_orders = spark.table('gizmobox.silver.py_orders')
display(df_orders)
```

```python
from pyspark.sql.functions import countDistinct, sum
df_summary = df_orders.groupBy('customer_id').agg(
    countDistinct('order_id').alias('total_orders'),
    sum('total_amount').alias('total_amount'),
    sum('quantity').alias('total_quantity')
)
display(df_summary)
```

```python
df_summary.writeTo('gizmobox.gold.orders_summary').createOrReplace()
```
