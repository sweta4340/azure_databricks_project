```python
df = spark.table('gizmobox.bronze.py_addresses')
display(df)
```

```python
from pyspark.sql.functions import first

df_pivot = (
  df.groupBy("customer_id").pivot("address_type")
    .agg(
      first("address_line_1").alias("address_line_1"),
      first("city").alias("city"),
      first("state").alias("state"),
      first("postcode").alias("postcode")
    )
)
display(df_pivot)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
df_pivot = df_pivot.withColumn("customer_id", col("customer_id").cast(LongType()))
```

```python
df_pivot.writeTo('gizmobox.silver.py_addresses').createOrReplace()
```
