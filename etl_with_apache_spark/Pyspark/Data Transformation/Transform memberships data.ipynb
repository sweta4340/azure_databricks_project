```python
df = spark.table('gizmobox.bronze.py_memberships')
display(df)
```

```python
# extract customer_id from path
from pyspark.sql.functions import *
df = df.withColumn('id', split(col('path'),'/')[7]).withColumn('customer_id',split(col('id'),'\\.')[0]).drop('id')
display(df)
```

```python
df.select('customer_id',col('content').alias('membership_card')).writeTo('gizmobox.silver.py_memberships').createOrReplace()
```
