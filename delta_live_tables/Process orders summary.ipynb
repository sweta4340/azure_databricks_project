```sql
%sql
CREATE OR REFRESH MATERIALIZED VIEW gold_orders_summary
AS
SELECT
  c.customer_id,
  c.customer_name,
  c.date_of_birth,
  c.telephone,
  c.email,
  a.address_line_1,
  a.city,
  a.state,
  a.postcode,
  count(distinct o.order_id) as total_orders,
  sum(o.item_quantity) as total_items,
  sum(o.item_quantity * o.item_price) as total_revenue
FROM LIVE.silver_customers as c
JOIN LIVE.silver_orders as o ON c.customer_id = o.customer_id
join LIVE.silver_addressess as a ON c.customer_id = a.customer_id
WHERE a.__end_at is null
GROUP BY ALL;
```
